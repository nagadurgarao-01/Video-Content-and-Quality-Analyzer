```python
df = df.drop_duplicates()

df["Title"] = df["Title"].fillna("")
df["Channel"] = df["Channel"].fillna("")
df["Examples"] = df["Examples"].fillna("")
df["Analogies"] = df["Analogies"].fillna("")
df["Visuals"] = df["Visuals"].fillna("")
```

```python
text_columns = [
    "Title",
    "Explanation_Type",
    "Visuals",
    "Examples",
    "Analogies",
    "Real_Life",
    "Quality",
    "Creator_Takeaways"
]

for col in text_columns:
    df[col] = df[col].fillna("").astype(str)
```

```python
def duration_to_seconds(duration):

    parts = duration.split(":")

    if len(parts) == 2:
        minutes, seconds = map(int, parts)
        return minutes * 60 + seconds

    if len(parts) == 3:
        hours, minutes, seconds = map(int, parts)
        return hours * 3600 + minutes * 60 + seconds

    return None
```

```python
df["duration_seconds"] = df["Duration"].apply(duration_to_seconds)
```

```python
def convert_views(x):

    x = str(x).replace("+", "").strip()

    if x.endswith("M"):
        return float(x[:-1]) * 1_000_000

    if x.endswith("K"):
        return float(x[:-1]) * 1_000

    return float(x)
```

```python
df["views_numeric"] = df["Views"].apply(convert_views)
```

```python
df["combined_text"] = (
    "Title: " + df["Title"] +
    "\nExplanation: " + df["Explanation_Type"] +
    "\nVisuals: " + df["Visuals"] +
    "\nExamples: " + df["Examples"] +
    "\nAnalogies: " + df["Analogies"] +
    "\nApplications: " + df["Real_Life"] +
    "\nQuality: " + df["Quality"] +
    "\nTakeaway: " + df["Creator_Takeaways"]
)
```

```python
print(df["combined_text"].iloc[0])
```

```python
df["has_analogy"] = (
    df["Analogies"].str.len() > 10
).astype(int)
```

```python
df["has_example"] = (
    df["Examples"].str.len() > 10
).astype(int)
```

```python
df["has_visual"] = (
    df["Visuals"].str.len() > 10
).astype(int)
```

```python
df["has_visual"] = (
    df["Visuals"].str.len() > 10
).astype(int)
```

```python
df["has_math"] = df["combined_text"].str.contains(
    r"formula|equation|mathematical|calculus|derivation|geometry|gradient",
    case=False,
    regex=True
).astype(int)
```

```python
df["has_code"] = df["combined_text"].str.contains(
    r"code|python|pytorch|tensorflow|programming|terminal|implementation",
    case=False,
    regex=True
).astype(int)
```

```python
df["has_application"] = (
    df["Real_Life"].str.len() > 10
).astype(int)
```
